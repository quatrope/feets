```python
# wily_api.py

import attrs
import contextlib
import io
from pathlib import Path

import wily
import pandas as pd

from wily.commands import graph, report, list_metrics
from wily.config import DEFAULT_CONFIG, WilyConfig

def _get_metrics() -> pd.DataFrame:
    rows = []
    for name, op in list_metrics.ALL_OPERATORS.items():
        for metric in op.operator_cls.metrics:
            rows.append({
                "Family": name,
                "Metric": metric.name,
                "Description": metric.description
            })
    return pd.DataFrame(rows)

def _get_metrics_names(metrics) -> frozenset:
    def join_metric(row):
        return f"{row.Family}.{row.Metric}"
    return frozenset(metrics.apply(join_metric, axis="columns").tolist())

METRICS = _get_metrics()
METRICS_NAMES = _get_metrics_names(METRICS)

del _get_metrics, _get_metrics_names


@attrs.define(frozen=True)
class WilyAPI:
    """Class to handle Wily metrics analysis on Python code."""

    config:  WilyConfig = attrs.field(default=DEFAULT_CONFIG, repr=True)
    
    
    def analyze_file(self, file_path: str, metrics: list|tuple=None) -> pd.DataFrame:
        metrics = METRICS_NAMES if metrics is None else tuple(metrics) 
        
        params = {
            "config": self.config,
            "path": str(file_path),
            "metrics": self.all_metrics,
            "n": None,
            "output": "/dev/null",
            "console_format": "html",
        }
        
        # Capture the output
        buff = io.StringIO()
        with contextlib.redirect_stdout(buff):
            report.report(**params)
        
        # Reset buffer and read results
        buff.seek(0)
        df = pd.read_html(buff)[0]
        
        # Clean metric values
        def clean_metric(v):
            return v.split()[0].strip()
        
        for column in df.columns[3:]:
            df[column] = df[column].apply(clean_metric)
            
        return df
```
